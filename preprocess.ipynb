```python
# Fill the missing values in the 'english_exam_score' column with the mean of the column
df["english_exam_score"] = df["english_exam_score"].fillna(df["english_exam_score"].mean())
```

```python
# Fill the missing values in the 'internship_duration_months' column with the median of the column for each 'internship_count' group
df['internship_duration_months'] = df.groupby('internship_count')['internship_duration_months'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df
```

```python
# Fill the missing values in the 'linkedin_profile_score' column with the mean of the column
df["linkedin_profile_score"] = df["linkedin_profile_score"].fillna(
    df["linkedin_profile_score"].mean()
)
```

```python
# Fill the missing values in the 'hr_interview_score' column with the mean of the column
df["hr_interview_score"] = df["hr_interview_score"].fillna(
    df["hr_interview_score"].mean()
)
```

```python
# Fill the missing values in the 'portfolio_score' column with the mean of the column
df['portfolio_score'] = df['portfolio_score'].fillna(
    df['portfolio_score'].mean()
    )
```

```python
# Fill the missing values in the 'github_avg_stars' column with the median of the column
df["github_avg_stars"]= df["github_avg_stars"].fillna(df["github_avg_stars"].median())

# Fill the missing values in the 'open_source_contribution_count' column with the median of the column
df["open_source_contribution_count"] = df["open_source_contribution_count"].fillna(
    df["open_source_contribution_count"].median())
```

```python
# Lasso ile seçilmiş feature listesi
selected_features = [
    "real_client_project_count",
    "cgpa",
    "hackathon_awards",
    "project_quality_score",
    "hackathon_count",
    "github_repo_count",
    "internship_duration_months",
    "technical_interview_score",
    "bootcamp_count",
    "certification_count",
    "communication_score",
    "problem_solving_score",
    "interviews_attended",
    "portfolio_score",
    "cloud_score",
    "teamwork_score",
    "coding_score",
    "github_avg_stars",
    "devops_score",
    "cv_quality_score",
    "open_source_contribution_count",
    "backend_score",
    "hr_interview_score",
    "presentation_score",
    "linkedin_profile_score",
    "leadership_score",
    "attendance_rate",
    "english_exam_score",
    "sql_score",
    "machine_learning_score",
    "applications_sent",
    "data_structures_score",
    "graduation_year",
    "target_role",
    "application_year",
    "failed_courses_count",
    "university_tier"
]

# Eğitim verisi
X_train = train_df[selected_features]
y_train = train_df["career_success_score"]

# Test verisi (target yok)
X_test = test_df[selected_features]
```

```python
# --- Model 1: Random Forest ---
rf_model = RandomForestRegressor(
    n_estimators=200, random_state=42, max_depth=8
)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict(X_test)
```

```python
# --- Model 2: Linear Regression ---
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)
lr_preds = lr_model.predict(X_test)
```

```python
# --- Tahminleri CSV formatına çevir ---
output_rf = pd.DataFrame({
    "student_id": student_id,
    "career_success_score": rf_preds
})

output_lr = pd.DataFrame({
    "student_id": student_id,
    "career_success_score": lr_preds
})

# Dosyaları kaydet
output_rf.to_csv("predictions_rf.csv", index=False)
output_lr.to_csv("predictions_lr.csv", index=False)
```
